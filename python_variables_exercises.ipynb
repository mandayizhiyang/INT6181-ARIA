# Beginner Python Exercise: Variables and `print()`

Create variables and use `print()` to display each story sentence. Try solving each exercise before looking at the example solution.

## Exercise 1: The Young Explorer
Create variables for a person's name and age. Print: `Mia is 10 years old and dreams of becoming an explorer.`

````python
name = "Mia"
age = 10

print(name, "is", age, "years old and dreams of becoming an explorer.")
````

## Exercise 2: A Magical Pet
Create variables for a pet's name and animal type. Print: `Luna is a magical cat who can glow in the dark.`

````python
pet_name = "Luna"
animal = "cat"

print(pet_name, "is a magical", animal, "who can glow in the dark.")
````

## Exercise 3: The Treasure Map

## Exercise 7: The Time Machine

## Exercise 9: The Mini Bakery
Create variables for the number of cupcakes and the price of each cupcake. Calculate the total price and print: `The bakery sold 12 cupcakes for a total of 24 dollars.`

````python
cupcakes = 12
price_each = 2
total_price = cupcakes * price_each

print("The bakery sold", cupcakes, "cupcakes for a total of", total_price, "dollars.")
````
